```python
X = df.drop(columns=["G3", "G1", "G2"])
y = df["G3"]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)

print("Original features:", X.shape)
print("Encoded features:", X_encoded.shape)
```

```python
X_encoded.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

linear_model.fit(X_train, y_train)
```

```python
linear_pred = linear_model.predict(X_test)

linear_mae = mean_absolute_error(y_test, linear_pred)
linear_rmse = np.sqrt(mean_squared_error(y_test, linear_pred))
linear_r2 = r2_score(y_test, linear_pred)

print("Linear Regression MAE:", linear_mae)
print("Linear Regression RMSE:", linear_rmse)
print("Linear Regression R²:", linear_r2)
```

```python
feature_importance = pd.Series(
    model.feature_importances_,
    index=X_encoded.columns
).sort_values(ascending=False)

feature_importance.head(15)
```

```python

plt.figure(figsize=(10, 6))

feature_importance.head(15).sort_values().plot(kind="barh")

plt.title("Top 15 Factors Used by the Random Forest Model")
plt.xlabel("Feature Importance")
plt.ylabel("Feature")

plt.show()
```

```python
model_comparison = pd.DataFrame({
    "Model": ["Random Forest", "Linear Regression"],
    "MAE": [mae, linear_mae],
    "RMSE": [rmse, linear_rmse],
    "R2 Score": [r2, linear_r2]
})

model_comparison
```
